You are designing a system to record and compare responses to a 10-question yes/no personality quiz. Each answer is either 1 (yes) or 0 (no). Use a single short int to store all 10 answers as bits.<br>

 

Implement the following functions:<br>

void setAnswer(CompWord* profile, int index, int value) – Set the answer to 1 or 0 for a given question index.<br>
int getAnswer(CompWord profile, int index) – Retrieve the answer for a given question index.<br>
int countYesAnswers(CompWord profile) – Count how many answers are 1.<br>
void printAnswers(CompWord profile) – Print all answers as a binary string.<br>
int compareProfiles(CompWord a, CompWord b) – Compares how many matching answers the two respondents have.<br>
CompWord unionProfiles(CompWord a, CompWord b) – Returns a new set containing all of the yes answers from both profiles.<br>
CompWord intersectProfiles(CompWord a, CompWord b) – Returns a new set containing all of the common yes answers from both profiles.<br>
CompWord differenceProfiles(CompWord a, CompWord b) – Returns a new set containing the yes answers from a and the common answers they have with b.<br>

You are designing a system to record and compare responses to a 10-question yes/no personality quiz. Each answer is either 1 (yes) or 0 (no). Use an array of char to store each answer.<br>

 

Implement the following functions:<br>

void setAnswer(BitVect* profile, int index, int value) – Set the answer to 1 or 0 for a given question index.<br>
int getAnswer(BitVect profile, int index) – Retrieve the answer for a given question index.<br>
int countYesAnswers(BitVect profile) – Count how many answers are 1.<br>
void printAnswers(BitVect profile) – Print all answers as a binary string.<br>
int compareProfiles(BitVect a, BitVect b) – Compares how many matching answers the two respondents have.<br>
BitVect unionProfiles(BitVect a, BitVect b) – Returns a new set containing all of the yes answers from both profiles.<br>
BitVect intersectProfiles(BitVect a, BitVect b) – Returns a new set containing all of the common yes answers from both profiles.<br>
BitVect differenceProfilesWord(BitVect a, BitVect b) – Returns a new set containing the yes answers from a and the common answers they have with b.<br>

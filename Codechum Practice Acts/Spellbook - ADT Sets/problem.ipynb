Objective: Implement an ADT for sets of strings in C to manage a wizard’s spellbook. Each spell is a unique string, and the spellbook must support typical set operations. The implementation is linked list.<br>

 

Create a C program that defines and uses a Set ADT with the following operations:<br>

Set* createSet() – Initializes an empty set.<br>
bool addSpell(Set*, const char*) – Adds a spell to the set (no duplicates).<br>
bool removeSpell(Set**, const char*) – Removes a spell from the set.<br>
bool containsSpell(Set*, const char*) – Checks if a spell exists in the set.<br>
void printSet(Set*) – Displays all spells in the set.<br>
void destroySet(Set* set) - Free all memory associated with the set.<br>
Set* unionSets(Set*, Set*) – Returns a new set containing all spells from both sets.<br>
Set* intersectSets(Set*, Set*) – Returns a new set containing only common spells.<br>
